```python
oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The Kings Speech",
    "The Artist",
    "Argo" 
]
years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232]

plt.plot(years,oscar_revenue)
plt.title("Oscar movies Revenue in each year")
plt.xlabel("Years")
plt.ylabel("Revenue ( in $M )")
plt.show()
```

```python
non_oscar_movies = ["Slumdog Millionaire", "Avatar", "Inception", "Hugo", "Lincoln"]
non_oscar_revenue = [378, 2788, 829, 185, 275] 

plt.plot(years,oscar_revenue, "o--r", label = "Oscar movies")
plt.plot(years,non_oscar_revenue, color = "y", marker = "<", linestyle = "-.", linewidth = 5,label = "Non-Oscar movies")

plt.title("Oscar VS Non-Oscar Movies Revenue")
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")
plt.legend(loc = "upper right") # default
plt.show()
```

```python
plt.plot(years, oscar_revenue, label = "Oscar movies")
plt.plot(years, non_oscar_revenue, label = "Non-Oscar movies")
plt.title("Oscar VS Non-Oscar Movies Revenue")
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")
plt.legend()
print(plt.style.available)
plt.style.use('dark_background')
plt.grid()
#plt.style.use('ggplot')
#plt.style.use("default")
plt.savefig("final_plot.png")
```

```python
movie = [
    "The Dark Knight",
    "The Hurt Locker",
    "The Kings Speech",
    "The Artist",
    "Argo" 
]
revenue = [1005, 170, 427, 133, 232]

plt.style.use("default")
plt.plot(movie, revenue)
plt.title("Movie Revenue")
plt.xlabel("Movies")
plt.ylabel("Revenue ($M)")
plt.tight_layout()
```

```python
# Bar Chart
plt.bar(years, revenue, color = "purple")
plt.title("Revenue in each year for Oscar movies")
plt.xlabel("Year")
plt.ylabel("Revenue in ($M)")
plt.ylim(0, max(revenue)+100)
for i in range(len(revenue)):
    plt.text(years[i], revenue[i]+10, str(revenue[i]), ha = "center")
plt.show()
```

```python
import numpy as np
oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The Kings Speech",
    "The Artist",
    "Argo" 
]
oscar_revenue = [1005, 170, 427, 133, 232]

non_oscar_movies = ["Slumdog Millionaire", "Avatar", "Inception", "Hugo", "Lincoln"]
non_oscar_revenue = [378, 2788, 829, 185, 275] 

years = [2008, 2009, 2010, 2011, 2012]

width = 0.4
x = np.arange(len(years))

plt.bar(x-width/2, oscar_revenue, width, label = "Oscar movies") # shifting to LEFT
plt.bar(x+width/2, non_oscar_revenue, width, label = "Non-Oscar movies") # shifting to RIGHT
plt.title("Oscar VS Non-Oscar movie revenue")
plt.xlabel("Year")
plt.ylabel("Revenue")
plt.legend()
plt.xticks(x,years)
plt.show()
```

```python
plt.barh(movie,revenue)
plt.xlabel("Revenue")
plt.ylabel("Movie")
plt.show()
```

```python
# Scatter Plots
people = ["Person A", "Person B", "Person C", "Person D", "Person E", "Person F", "Person G", "Person H", "Person I", "Person J"]
age = [22,25,30,35,40,45,50,55,60,65]
bp = [110,115,120,122,125,130,135,123,145,150] 

#colors = ["red", "yellow", "pink", "green", "purple", "red", "yellow", "pink", "green", "purple"]
colors = ["green" if x < 135 else "red" for x in bp]

#plt.scatter(age, bp, color = colors, s = bp, marker = "o", alpha = 0.2)
plt.scatter(age, bp, s = bp, cmap = "plasma", c = bp)
plt.title("AGE VS BP")
plt.xlabel("Age")
plt.ylabel("BP")
plt.grid()
plt.colorbar(label = "BP")

for i in range(len(people)):
    plt.annotate(people[i], xy = (age[i], bp[i]), xytext = (age[i]+1, bp[i]))
plt.xlim(min(age)-10, max(age)+10)
plt.show()
```

```python
cities = ["city A", "city B", "city C", "city D", "city E"]

winter_temp = [5,2,10,0,7]
winter_humidity = [80,75,65,85,70]

summer_temp = [25,30,28,35,27]
summer_humidity = [60,50,55,45,65]

plt.scatter(winter_temp, winter_humidity, label = "Winter")
plt.scatter(summer_temp, summer_humidity, label = "Summer", color = "orange")
plt.title("Summer VS Winter Data")
plt.xlabel("Temperature (°C)")
plt.ylabel("Humidity (%)")
plt.legend()
plt.show()
```

```python
# Pie Charts
expenses = ["salaries", "rent", "marketing", "R&D", "Miscellaneous"]
amounts = [500,150,120,100,50]
colors = ["red", "black", "pink", "green", "blue"]
explode = [0,0,0,0.2,0]
plt.pie(amounts, labels = expenses, autopct="%1.1f%%", colors = colors, wedgeprops = {
    "edgecolor" : "black", 
    "linewidth" : 1,
    "linestyle" : "--"
    # "alpha" : 0.8
    # "fill" : False
    }, explode = explode, startangle = 90, shadow = True)
print(plt.style.available)

plt.style.use("default")
plt.show()
```
