MELT AND PIVOT

melt - wide to long format
melt (id_vars, value_vars, var_name, value_name) # keep, unpivot, new_col_name, new_col_value


pivot - long to wide format
pivot (index, column, values)
      # keep

```python
df = pd.DataFrame ({
    "country":["USA", "USA", "India", "India"],
    "year":[2020,2021,2020,2021],
    "sales":[100,120,90,110],
    "profit":[20,25,18,22],
    "tax":[10,20,30,40]
})
df
```

```python
melted_data = df.melt(
    id_vars = ["country","year"],
    value_vars =  ["sales","profit","tax"],
    var_name = "metrics",
    value_name = "value"
)
melted_data
```

```python
original = melted_data.pivot(
    index = ["country","year"],
    columns = "metrics",
    values = "value"
)
original
```

```python
# MERGING AND JOINING THE DATA

df_customers = pd.DataFrame({
    "customer_id" : [1,2,3,4],
    "name" : ["Adam", "Bob", "Charlie", "Dave"]
})

df_orders = pd.DataFrame({
    "order_id" : [101,102,103,104],
    "customer_id" : [2,1,5,4],
    "amount" : [250,120,300,180]
})
```

```python
df_customers
```

```python
df_orders
```

```python
pd.merge(df_customers, df_orders, on = "customer_id") # inner join
```

```python
pd.merge(df_customers, df_orders, on = "customer_id", how = "left") # left join
```

```python
pd.merge(df_customers, df_orders, on = "customer_id", how = "right") # right join
```

```python
pd.merge(df_customers, df_orders, on = "customer_id", how = "outer") # outer join
```
