```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.set_theme()
```

```python
sns.get_dataset_names()
```

```python
tips = sns.load_dataset("tips")
print(type(tips))
```

```python
print(tips.head())
```

```python
# relational plots (show relationship) - scatter and line plots
sns.relplot(
    data = tips,
    x = "total_bill",
    y = "tip",
    hue = "smoker",
    size = "size",
    style = "smoker"
)
```

```python
sns.set_theme(style = "ticks")
sns.relplot(
    data = tips,
    x = "total_bill",
    y = "tip",
    kind = "line"
)
```

```python
# Categorical plots (analyze on basis of category) - bar plot and box plot
print(tips.head())
```

```python
sns.barplot(
    data = tips,
    x = "day",
    y = "tip",
    hue = "sex"
)
```

```python
sns.boxplot(
    data = tips,
    x = "day",
    y = "tip"
)
```

```python
# Distribution plots - Histogram
penguins = sns.load_dataset("penguins")
penguins.head()
```

```python
sns.histplot(
    data = penguins,
    x = "body_mass_g",
    bins = 30
)
```

```python
# Matrix plots - Heatmaps (plot rectangular data as color encoded matrix)
flights = sns.load_dataset("flights")
flights.head()
```

```python
flights = flights.pivot(index ="month" , columns ="year" , values ="passengers" ) # long to wide
```

```python
sns.heatmap(
    flights,
    cmap = "coolwarm",
    annot = True,
    fmt = "d"
)
plt.title("Passengers Heatmap")
plt.tight_layout()
```

```python
fig,ax = plt.subplots()

sns.lineplot(
    data = tips,
    x = "day",
    y = "total_bill",
    hue = "sex",
    ax = ax,
    marker = "o",
    errorbar = None
)
ax.set_title("Avg bill value for each day")
ax.set_xlabel("Days")
ax.set_ylabel("Total bill amount")
fig.tight_layout()
```
